## 📊 Example 1: Simple WordCount

Let's count words in a text sample:

```python
import apache_beam as beam
from apache_beam.options.pipeline_options import PipelineOptions

# Sample data
quotes = [
    'To be or not to be that is the question',
    'Be yourself everyone else is already taken',
    'Two things are infinite the universe and human stupidity',
]

# Create and run pipeline
with beam.Pipeline(options=PipelineOptions()) as pipeline:
    word_counts = (
        pipeline
        | 'Create' >> beam.Create(quotes)
        | 'Split' >> beam.FlatMap(lambda x: x.lower().split())
        | 'PairWithOne' >> beam.Map(lambda x: (x, 1))
        | 'GroupAndSum' >> beam.CombinePerKey(sum)
    )
    
    # Print results
    results = []
    def collect(element):
        results.append(element)
    
    word_counts | beam.Map(collect)

# Display top 10 words
print('\n🎯 Top 10 Most Frequent Words:')
print('-' * 40)
sorted_results = sorted(results, key=lambda x: x[1], reverse=True)[:10]
for word, count in sorted_results:
    print(f'{word:.<25} {count}')
```

## 🔥 Example 2: Advanced Transform with Custom DoFn

```python
class FilterLongWords(beam.DoFn):
    """Keep only words longer than 5 characters."""
    def process(self, element):
        word, count = element
        if len(word) > 5:
            yield {'word': word, 'count': count, 'length': len(word)}

with beam.Pipeline(options=PipelineOptions()) as pipeline:
    long_words = (
        pipeline
        | 'CreateData' >> beam.Create(results)
        | 'FilterLong' >> beam.ParDo(FilterLongWords())
        | 'Print' >> beam.Map(print)
    )

print('\n✅ Filtered words longer than 5 characters!')
```

## 📈 Example 3: Real-time Analytics Simulation

Simulate streaming data with windowing:

```python
import time
from datetime import datetime

# Simulate streaming events
events = [
    {'user_id': 'user1', 'action': 'click', 'timestamp': time.time()},
    {'user_id': 'user2', 'action': 'view', 'timestamp': time.time()},
    {'user_id': 'user1', 'action': 'purchase', 'timestamp': time.time()},
    {'user_id': 'user3', 'action': 'click', 'timestamp': time.time()},
    {'user_id': 'user2', 'action': 'click', 'timestamp': time.time()},
]

with beam.Pipeline(options=PipelineOptions()) as pipeline:
    user_actions = (
        pipeline
        | 'CreateEvents' >> beam.Create(events)
        | 'ExtractUser' >> beam.Map(lambda x: (x['user_id'], x['action']))
        | 'CountPerUser' >> beam.combiners.Count.PerKey()
        | 'Format' >> beam.Map(lambda x: f"User {x[0]} performed {x[1]} actions")
        | 'Display' >> beam.Map(print)
    )

print('\n✅ User activity analytics complete!')
```
